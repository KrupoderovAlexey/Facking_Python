```python
# Чтение входных данных
matrix = []
while True:
    line = input()
    if line == 'end':
        break
    matrix.append([int(x) for x in line.split()])

# Определение размеров матрицы
# ... (чтение матрицы) ...

rows = len(matrix)
cols = len(matrix[0]) # Исправлено: считаем именно столбцы

for i in range(rows):
    row_result = []
    for j in range(cols):
        # Используем скобки для корректного циклического сдвига
        val_up = matrix[i - 1][j]
        val_down = matrix[(i + 1) % rows][j]
        val_left = matrix[i][j - 1]
        val_right = matrix[i][(j + 1) % cols] # Исправлено: скобки
        
        row_result.append(val_up + val_down + val_left + val_right)
    
    print(*row_result)
```

```python
n = int(input())

# Создаем пустую матрицу n x n
matrix = [[0] * n for _ in range(n)]

num = 1
left, right = 0, n - 1
top, bottom = 0, n - 1

while num <= n * n:
    # Идем вправо
    for i in range(left, right + 1):
        matrix[top][i] = num
        num += 1
    top += 1
    
    # Идем вниз
    for i in range(top, bottom + 1):
        matrix[i][right] = num
        num += 1
    right -= 1
    
    # Идем влево
    for i in range(right, left - 1, -1):
        matrix[bottom][i] = num
        num += 1
    bottom -= 1
    
    # Идем вверх
    for i in range(bottom, top - 1, -1):
        matrix[i][left] = num
        num += 1
    left += 1

# Вывод матрицы
for row in matrix:
    print(*(row))
```

```python
def update_dictionary(d, key, value):
    if key in d:
        d[key].append(value)
    else:
        new_key = 2 * key
        if new_key in d:
            d[new_key].append(value)
        else:
            d[new_key] = [value]

d = {}
print(update_dictionary(d, 1, -1))  # None
print(d)                            # {2: [-1]}
update_dictionary(d, 2, -2)
print(d)                            # {2: [-1, -2]}
update_dictionary(d, 1, -3)
print(d)                            # {2: [-1, -2, -3]}
```

```python
n = int(input())
s = []
for _ in range(n):
    line = input().strip()
    parts = [x.strip() for x in line.split(';')]
    s.append(parts)

# Словарь для хранения статистики по каждой команде
stats = {}

for match in s:
    team1, score1, team2, score2 = match[0], int(match[1]), match[2], int(match[3])
    
    # Инициализируем статистику для команд, если их еще нет в словаре
    if team1 not in stats:
        stats[team1] = {'games': 0, 'wins': 0, 'draws': 0, 'losses': 0, 'points': 0}
    if team2 not in stats:
        stats[team2] = {'games': 0, 'wins': 0, 'draws': 0, 'losses': 0, 'points': 0}
    
    # Увеличиваем счетчик игр для обеих команд
    stats[team1]['games'] += 1
    stats[team2]['games'] += 1
    
    # Определяем результат матча
    if score1 > score2:
        # Первая команда выиграла
        stats[team1]['wins'] += 1
        stats[team1]['points'] += 3
        stats[team2]['losses'] += 1
    elif score1 == score2:
        # Ничья
        stats[team1]['draws'] += 1
        stats[team1]['points'] += 1
        stats[team2]['draws'] += 1
        stats[team2]['points'] += 1
    else:
        # Вторая команда выиграла
        stats[team1]['losses'] += 1
        stats[team2]['wins'] += 1
        stats[team2]['points'] += 3

# Выводим результаты
for team, data in stats.items():
    print(f"{team}:{data['games']} {data['wins']} {data['draws']} {data['losses']} {data['points']}")
```
